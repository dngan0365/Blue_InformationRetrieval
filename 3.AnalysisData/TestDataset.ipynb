```python
import csv
import os

# Tạo folder kết quả
output_folder = "test/query_results"
os.makedirs(output_folder, exist_ok=True)

# Đọc dữ liệu URLs từ file CSV thứ hai
url_data = {}
with open("../3.AnalysisData/output/final_document_tfidf_pagerank.csv", newline="", encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)
    for row in reader:
        url_data[row["id"]] = row["url"]

# Hàm tạo file txt cho mỗi query
def create_query_files(query_csv, keep_slash=True):
    with open(query_csv, newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            stt = row["stt"]
            # Lấy danh sách URL từ file queries.csv
            import json
            urls = json.loads(row["urls"])
            
            # Xử lý / nếu cần
            if not keep_slash:
                urls = [u.rstrip("/") for u in urls]

            # Lấy id tương ứng từ url_data
            lines = []
            for url in urls:
                # Tìm id tương ứng
                found_ids = [id_ for id_, u in url_data.items() if u.rstrip("/") == url.rstrip("/")]
                for id_ in found_ids:
                    lines.append(f"{id_} {url}")

            # Ghi vào file stt.txt
            output_file = os.path.join(output_folder, f"{stt}.txt")
            with open(output_file, "w", encoding="utf-8") as out_f:
                out_f.write("\n".join(lines))
            print(f"✅ Tạo file: {output_file}")

# Gọi hàm
create_query_files("test/queries.csv", keep_slash=False)
```
